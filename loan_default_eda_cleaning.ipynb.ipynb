```python
# Filling the missing numbers with the median (middle value), and missing text with the most common value
df['Income'] = df['Income'].fillna(df['Income'].median())
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())
df['Education'] = df['Education'].fillna(df['Education'].mode()[0])
```

Check and remove duplicate

```python
df.duplicated().sum()        # counting exact duplicate rows
```

Distribution and Outlier Check

```python
# Look at the lowest and highest values to spot anything unrealistic
print(df['Income'].describe())
print(df['LoanAmount'].describe())
```

Feature Engineering

```python
#Step 1: converting yearly interest rate into a monthly rate
monthly_rate = (df['InterestRate'] / 100) / 12
```

```python
#Step 2: calculate EMI (monthly loan payment) using the standard EMI formula
df['EMI'] = (df['LoanAmount'] * monthly_rate * (1 + monthly_rate) ** df['LoanTerm']) / \
            ((1 + monthly_rate) ** df['LoanTerm'] - 1)
```

```python
# Step 3: EMI as a % of monthly income — shows how much of income goes to loan payment
df['EMI_to_Income'] = (df['EMI'] / (df['Income'] / 12)).round(3)
```

```python
# Step 4: group numeric values into easy-to-read bands/categories
df['CreditScoreBand'] = pd.cut(df['CreditScore'], bins=[0, 579, 669, 739, 850],
                                labels=['Poor', 'Fair', 'Good', 'Excellent'])
df['DTIBand'] = pd.cut(df['DTIRatio'], bins=[0, 0.2, 0.4, 0.6, 1],
                        labels=['Low', 'Moderate', 'High', 'Very High'])
```

```python
# Step 5: a simple flag for job stability
df['TenureStability'] = np.where(df['MonthsEmployed'] >= 24, 'Stable', 'Unstable')
```

#Correlation checking

```python
# Simple correlation table — values closer to 1 or -1 mean a stronger relationship
cols = ['Income', 'LoanAmount', 'CreditScore', 'InterestRate', 'DTIRatio', 'EMI_to_Income', 'Default']
print(df[cols].corr()['Default'].sort_values(ascending=False))
```
